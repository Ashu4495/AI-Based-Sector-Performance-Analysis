# MarketPulse AI — Exploratory Data Analysis (EDA)
Analysis of 8 NSE Sectors vs. NIFTY 50 Benchmark
- Features: Momentum, Volatility, Moving Averages, Relative Strength
- Distributions & Sector Rotation Profiles

```python
import pandas as pd
import numpy as np
from backend.app.core.config import SECTORS, BENCHMARK
from backend.app.features.engineer import load_feature_store, ALL_FEATURE_COLUMNS

df = load_feature_store()
print(f"Loaded {len(df)} feature rows across {df['Sector'].nunique()} sectors.")
df.head()
```

## 1. Sector Volatility & Risk-Return Metrics

```python
stats = []
for sector in df['Sector'].unique():
    sec_df = df[df['Sector'] == sector]
    stats.append({
        'Sector': sector,
        'Name': SECTORS.get(sector, {}).get('name', sector),
        'Avg Volatility (Ann %)': round(sec_df['volatility_20d'].mean() * 100, 2),
        'Avg 20d Return (%)': round(sec_df['return_20d'].mean() * 100, 2),
        'Max 20d Return (%)': round(sec_df['return_20d'].max() * 100, 2),
        'Min 20d Return (%)': round(sec_df['return_20d'].min() * 100, 2)
    })
pd.DataFrame(stats)
```
